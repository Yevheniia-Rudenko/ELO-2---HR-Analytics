# Data cleaning
This notebook clean and process the raw data to prepare them for analysis

## Import libraries

```python
import pandas as pd
```

## Load raw data

```python
df = pd.read_csv("..//1_datasets/HR-Employee-Attrition-raw.csv")
print(df.shape)
#df.info()
#df.describe()
df.head()
```

## Data cleaning

Check if there are any fields that contain NaN/null values

```python
print(df.isnull().sum())
```

Not all of the fields contain completed entries. It means, that some information is missed. The possible reasons of missed information are:
1. new employees have not submited surveys,
2. managers have not provided feedback for new employees
3. discrepancies in the data of the information system

Select only lines with NaN values and analyse them

```python
nan_rows = df[df.isnull().any(axis=1)]
print(nan_rows.shape)
print(nan_rows.info())
nan_rows.head()
```

1. New employees have not submited surveys.
Periodically employees complete surveys ragarding their satisfaction or engagement in the work. There could be some new employees, who have not provided this information. The source of the employee's survey have the following filds: EnvironmentSatisfaction, JobSatisfaction, RelationshipSatisfaction, WorkLifeBalance.
Check if there are new employees who have not provided their surveys yet.

```python
missed_survey_nan = df[df['JobSatisfaction'].isnull()]
print(missed_survey_nan.shape)
print(missed_survey_nan.info())
missed_survey_nan.head()
```

There are 3 employees who worked in the company 0 years (field"YearsAtCompany") and have missed ('NaN') values for the fields: 'EnvironmentSatisfaction', 'JobSatisfaction', 'RelationshipSatisfaction', 'WorkLifeBalance'. These fields are poplated from the periodical surveys. Since the employees are new, they have not completed these surveys yet.

2. Managers have not provided feedback for new employees.
Periodically managers provide assessment of their suboridnates. In the case of new employees this assessment can be missed. The fileds: 'JobInvolvement', 'PerformanceRating' come from periodical manageres assesments.
Check if there are new employees who have not obtained manager's assesment yet.

```python
missed_manager_assesment_nan = df[df['JobInvolvement'].isnull()]
print(missed_manager_assesment_nan.shape)
print(missed_manager_assesment_nan.info())
missed_manager_assesment_nan.head()
```

There are 2 employees who worked in the company 0 years (field"YearsAtCompany") and have missed ('NaN') values for the fields: 'JobInvolvement', 'PerformanceRating'. These fields are poplated from the periodical managers asesments. Since the employees are new, they have not abtained any manager's assessmant yet.

3. Discrepancies in the data of the information system
Not all the case (lines) that include blank values were explained by hipothesis 1 and 2. It means that there are other discrepancies in the data that have other reason of missed information (different from not completed surveys and missed manager's appraisal).

```python
selected_indices = missed_survey_nan.index.union(missed_manager_assesment_nan.index)
rest_discrep_nan = nan_rows[~nan_rows.index.isin(selected_indices)]
print(rest_discrep_nan.shape)
print(rest_discrep_nan.info())
rest_discrep_nan.head()
```

There is no logical explanation for this missing information. The reason could be a lack of information during data migration from previous systems, or missing input from the responsible HR specialist. For the purpose of data abalysis all these cases should be expluded from the raw dataset. Existance of cases with blank values may produce distorted results?, especcialy for the tasks of correlation analysis.

Remove all the lines that include 'NaN' values.

```python
df_cleaned = df.dropna()
print(df_cleaned.shape)
print(df_cleaned.info())
df_cleaned.head()
```

Save cleaned dataset in HR-Employee-Attrition.csv file

```python
df_cleaned.to_csv('..//2_data_preparation/HR-Employee-Attrition.csv', index=False)
```
